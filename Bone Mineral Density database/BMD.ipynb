We replace the missing weights of the males with the mean male weight and the missing weights of females with he mean female weight.

```python
df_male = df[df['sex'] == 'M']
male_weight = df_male['weight_kg'].mean()

df_female = df[df['sex'] == 'F']
female_weight = df_female['weight_kg'].mean()
```

```python
df.loc[(df['weight_kg'].isna()) & (df['sex'] == 'M'), 'weight_kg'] = male_weight

df.loc[(df['weight_kg'].isna()) & (df['sex'] == 'F'), 'weight_kg'] = female_weight
```

Three values look like data errors. Suggest a reasonable strategy to handle the data errors:


*   Delete rows
*   Replace with mean
*   Winsorize at 95%



1.implement deletion and replacement with mean

```python
df_delete_high_bmd = df[df['bmd'] < 10]
```

```python
plt.boxplot(df_delete_high_bmd['bmd'], vert=False)
plt.show()
```

2.replace high bmd with mean of correct values

```python
df['bmd_mean_replacement'] = df['bmd']

df.loc[df['bmd'] > 4, 'bmd_mean_replacement'] = df_delete_high_bmd['bmd'].mean()
```

3.winsorize at 95% for upper values

```python
upper_quantile = df['bmd'].quantile(0.95)
print(f'95% quantile is {upper_quantile}.')

df['bmd_winsorized'] = df['bmd'].clip(upper=upper_quantile)
df['bmd_winsorized'].describe()
```

Select the winsorized data as definite replacement strategy: set the bmd column to bmd_winsorized

```python
df["bmd"] = df["bmd_winsorized"]
```

Calculate the body mass index as the weight in kg divided by the squared height in meters. Add a new column "bmi".

```python
df['bmi'] = df['weight_kg'] / ((df['height_cm'] / 100) ** 2)

df['bmi'].describe()
```

Plot the mean mineral density of males and females using a bar plot.

```python
mean_male = df.loc[df['sex']=='M', 'bmd'].mean()
mean_female = df.loc[df['sex']=='F', 'bmd'].mean()

print(mean_male, mean_female)
```

```python
plt.figure(figsize=(4, 4))
plt.bar(["male", "female"], [mean_male, mean_female], width=0.2)

plt.title("Bone mineral density", fontsize=16)
plt.xlabel("Sex", fontsize=14)
plt.ylabel("BMD", fontsize=14)
plt.xticks(rotation=0, fontsize=12)
plt.yticks(rotation=0, fontsize=12)
plt.show()
```
